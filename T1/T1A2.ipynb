```python
import numpy as np
```

```python
def simpson_two_slices(f, x_i, h):
    """
    Simpson's rule over [x_i-h, x_i+h]: (h/3)(f_{i-1} + 4f_i + f_{i+1}).

    f:   f(x)
    x_i: center point of the 2 slices
    h:   space between nodes
    """
    return (h/3) * (f(x_i - h) + 4*f(x_i) + f(x_i + h))
```
