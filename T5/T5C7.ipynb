# Question T5C7: Detect nonsingular matrices

```python
import numpy as np
import random
nummax=7
N=6
A = np.zeros((N,N))
b = np.zeros((N,1))

for i in range(N):
    for j in range(N):
        A[i][j]=random.randint(-nummax,nummax)
        b[i][0]=random.randint(-nummax,nummax)

#The next two lines are used to create a singular matrix. Remove the comment signs to use the program with a random matrix.
        
A[[2]]=-A[[1]]
b[2]=-b[1]

print(np.matrix(A))
print(np.matrix(b))
```

I check if A is singular with Gaussian elimination with partial pivoting. If at some column even the largest value I can pick as pivot is 0, the rows are linearly dependent and the matrix is singular.

```python
def check_singular(A):
    U = np.array(A, dtype=float) #copy, so A is not changed
    n = U.shape[0]
    # Because of rounding errors a pivot is almost never exactly 0,
    # so I call it 0 when it is very small compared to the entries of A
    tol = 1e-10 * np.max(np.abs(U))

    for j in range(n):
        # Partial pivoting: row with the largest |U[i, j]| from row j down
        p = j + np.argmax(np.abs(U[j:, j]))
        U[[j, p]] = U[[p, j]]

        # Even the largest value in the column is 0, so this column has no pivot
        # Then the rows are linearly dependent and the matrix is singular
        if np.abs(U[j, j]) <= tol:
            print("The matrix is singular, so the system has no unique solution.")
            return True

        # Make the elements below the pivot 0
        for i in range(j + 1, n):
            m = U[i, j] / U[j, j]
            U[i, j:] -= m * U[j, j:]

    # Every column had a nonzero pivot, so the matrix is nonsingular
    print("The matrix is nonsingular, so the system has a unique solution.")
    return False
```

```python
singular = check_singular(A)
```
