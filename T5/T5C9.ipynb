# Determinant of an n x n matrix

Note:
- I use Gaussian elimination with partial pivoting to make the matrix upper triangular (section 5.3 in the book).
- Subtracting a multiple of one row from another row does not change the determinant.
- Swapping two rows changes the sign of the determinant, so I count the swaps.
- The determinant of an upper triangular matrix is the product of its diagonal elements.
- Therefore: det(A) = (-1)^(number of swaps) × (product of the diagonal elements after elimination).
- If the pivot is still 0 after pivoting, the whole column below the diagonal is 0. Then the matrix is singular and det(A) = 0, which comes out automatically because there is a 0 on the diagonal.

```python
#determinant with numpy library along with the matrix

import numpy as np


A=[ [1,2,1], [3,6,0], [3,5,0] ]
print("Determinant according to Numpy: {}".format(np.linalg.det(A)))
```

```python
# Determinant with Gaussian elimination and partial pivoting
# A: any n x n matrix
# Prints my determinant and the one from numpy, so I can check my algorithm
def determinant(A):
    U = np.array(A, dtype=float) #copy, so A is not changed
    n = U.shape[0]
    swaps = 0

    for j in range(n - 1):
        # Partial pivoting: row with the largest |U[i, j]| from row j down
        p = j + np.argmax(np.abs(U[j:, j]))
        if p != j:
            U[[j, p]] = U[[p, j]]
            swaps += 1 #every swap flips the sign of the determinant

        # Pivot is 0: the whole column below is 0, nothing to eliminate (det will be 0)
        if U[j, j] == 0:
            continue

        # Make the elements below the pivot 0 (this does not change the determinant)
        for i in range(j + 1, n):
            m = U[i, j] / U[j, j]
            U[i, j:] -= m * U[j, j:]

    # Upper triangular now: det = product of the diagonal, with the sign from the swaps
    det = (-1)**swaps * np.prod(np.diag(U))

    print("My determinant:                 ", det)
    print("Determinant according to Numpy: ", np.linalg.det(A))
    return det
```

```python
A = [[1, 2, 1],
     [3, 6, 0],
     [3, 5, 0]]
det = determinant(A)
```

Note:
- Without pivoting this matrix fails: after the first column is eliminated, the second pivot is 0. With pivoting it works.
- Check by hand, expanding along the third column: det(A) = 1 × (3·5 - 6·3) = -3.
- My result is exactly -3. Numpy gives -3.0000000000000004, the difference is only floating point rounding.
